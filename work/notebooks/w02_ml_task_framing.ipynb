## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

## My lane as an ML task

**Task type: Ranking**

I am framing Lane 2, Refresh / Content Opportunity Scoring, as a ranking task. The goal is to rank pages by their priority for human review rather than simply classify every page as good or bad.

The output would be an ordered list of pages that helps an SEO or content team decide which pages to investigate first. This matches the real decision because the team has limited review capacity and needs to prioritize its attention.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## Target or proxy

**Proxy target: `trend_direction`**

For the starter dataset, I will use the observed `trend_direction` field as a proxy for identifying pages that may deserve attention. In particular, pages with a `down` trend can be treated as a signal for further investigation.

This is a proxy rather than the ideal target. The stronger target for the eventual problem would be a future outcome, such as whether a page's performance declines during a defined period after the features are observed. The current starter data does not establish that future outcome, so I will not treat the existing trend label as proof that a page needs a specific content action.


## 3. Success metric

*One metric you can defend. What number means 'good'?*

## Success metric

The primary success metric will be **Precision@K**, where K represents the number of pages the team can realistically review.

For example, Precision@50 measures how many of the top 50 ranked pages correspond to the proxy outcome we are trying to surface. This metric fits the decision because the team has limited review capacity and cares about whether the highest-priority recommendations are useful.

I would also compare the ranking against a simple fixed-rule baseline to determine whether the ML approach provides meaningful improvement.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML beats a fixed rule here

A fixed rule could prioritize pages using a single signal, such as selecting every page with a `down` trend or pages that have not been updated for a certain number of days. This is easy to understand, but it may miss interactions between multiple signals.

An ML approach could consider several page-level features together, such as search volume, impressions, clicks, engagement, content age, average position, and recent trends. It could learn patterns in the starter data and use them to produce a ranked list rather than applying one fixed threshold.

The goal is not to replace human judgment. The goal is to improve how limited human review time is allocated by surfacing potentially higher-priority pages first.
