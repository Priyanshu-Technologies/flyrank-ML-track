## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## The question

**Research question:**  
Can observable page-level search, engagement, visibility, and content signals be used to prioritize which pages an SEO or content team should review first?

**Decision:**  
Which pages should the team prioritize for human review?

**Action:**  
A reviewer could investigate the highest-priority pages and decide whether each page should be refreshed, expanded, protected, pruned, or monitored.

**Cost of a wrong call:**  
A false positive could cause a reviewer to spend limited time investigating a page that does not need attention. A false negative could cause the team to miss a page with a meaningful opportunity or problem. Since review capacity is limited, the quality of the prioritization matters.

## Careful words: what I can and can't claim

### What I can claim

The starter dataset can show observed patterns and associations between page-level characteristics and the available performance signals. These signals can be investigated to determine whether they are useful for prioritizing pages for human review.

A future model could provide decision support by ranking pages according to patterns found in the available data.

### What I can't claim

I cannot claim that a page's observed trend is caused by any particular content characteristic, or that refreshing a page will definitely improve its performance. The starter data does not establish causation.

I also cannot claim that a model will predict Google's ranking decisions or guarantee future search performance. Any results from the starter dataset should be treated as evidence for further investigation rather than proof that the final approach will work.
